## Series de Tiempo


<!-- DEFINICIÓN -->
<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: serie temporal</strong>
Una **serie de tiempo** o **serie temporal** es una secuencia de observaciones de una variable registradas en distintos instantes y ordenadas cronológicamente. Por ejemplo, la temperatura medida cada hora en la estación de Curicó forma una serie temporal: cada observación tiene un valor y una fecha y hora asociadas.

El orden importa porque las observaciones pueden depender de las anteriores y presentar patrones que se repiten, como los cambios de temperatura entre el día y la noche. Al analizar una serie, buscamos comprender esa evolución y, al pronosticar, usar la información disponible para anticipar valores futuros.
</div>



### Pronóstico de una Serie de Tiempo

Al pronosticar una **serie temporal**, el orden de las observaciones determina qué información puede usarse para entrenar y evaluar. Cada predicción debe construirse con los datos que habrían estado disponibles al momento de emitirla. Si a las 18:00 queremos anticipar una helada durante la noche, podemos usar las mediciones registradas hasta esa hora, pero ninguna posterior, aunque aparezca en el mismo conjunto de datos.

Esta restricción también se aplica a la evaluación: debemos entrenar con un período anterior y evaluar con uno posterior, respetando el horizonte de pronóstico. Una partición aleatoria puede mezclar pasado y futuro, y terminar evaluando predicciones con un modelo que aprendió de observaciones posteriores a ellas.


<!-- DEFINICIÓN -->
<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: fuga de información (data leakage)</strong>
Hay **fuga de información** (*data leakage*) cuando el entrenamiento, la construcción de features o la selección del modelo utilizan información que no estaría disponible en la situación que queremos evaluar. En un pronóstico, ocurre, por ejemplo, al usar mediciones posteriores al origen, calcular una imputación con todo el período o elegir hiperparámetros mirando test. La fuga puede producir métricas demasiado optimistas: el modelo parece anticipar el futuro, pero recibió información que no tendría al operar.
</div>


### Origen, horizonte y target

Para definir qué debe predecir el modelo, necesitamos distinguir tres conceptos:


<!-- DEFINICIÓN -->
<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: origen, horizonte, target y features</strong>

- **Origen (`t`):** el instante en que se emite el pronóstico. Es el punto de referencia para determinar qué información podemos usar.
- **Horizonte (`h`):** cuánto tiempo hacia adelante queremos predecir. En este laboratorio lo expresaremos en horas y trabajaremos principalmente con un horizonte de **12 horas**.
- **Target (`y_t`):** la variable que queremos predecir para el ejemplo cuyo origen es `t`. En el problema horario, corresponde a la temperatura observada `h` horas después: `y_t = temperatura(t + h)`. Con el horizonte principal de 12 horas, usamos `y_t = temperatura(t + 12)`. Más adelante definiremos un segundo problema, cuyo target será la mínima observada durante una ventana nocturna.

Las **features** son las variables que el modelo utiliza para construir el pronóstico. Pueden incluir la temperatura en el origen, mediciones anteriores y resúmenes del pasado, como la temperatura media de las últimas horas. En este laboratorio, el origen puede ser cualquier hora: si `t` corresponde a las **18:00**, las features deben permitirnos pronosticar la temperatura de las **06:00 del día siguiente**.
</div>


La regla es que **cada feature debe poder calcularse con información disponible al emitir el pronóstico**. Supondremos que las mediciones se reciben sin retraso: en `t` podemos usar lo observado hasta esa hora, incluida la medición de `t`. También conocemos de antemano información de calendario, como la hora o el mes del instante que queremos pronosticar. En cambio, no podemos usar la temperatura observada en `t + 1` ni un promedio que incluya mediciones posteriores a `t`.

Aunque la tabla histórica contenga todas esas mediciones, debemos reconstruir cada ejemplo como si estuviéramos en su origen, sin haber visto lo que ocurrió después. **El target se conoce más tarde y permite evaluar el pronóstico; las features deben estar disponibles al momento de hacerlo.**

### Componentes de una serie

> Esta sección introduce algunos conceptos básicos de series temporales, pero no necesariamente la utilizaremos dentro del laboratorio.


<!-- DEFINICIÓN -->
<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: descomposición y componentes de una serie</strong>
La temperatura cambia a distintas escalas: baja durante la noche, varía entre invierno y verano y puede atravesar períodos más cálidos o fríos de lo habitual. La **descomposición de una serie temporal** permite distinguir esos patrones y examinar qué variación queda sin explicar.

Conceptualmente, podemos distinguir cuatro componentes: **tendencia**, **estacionalidad**, **ciclo** y **componente irregular**. En la práctica, muchos métodos agrupan la tendencia y el ciclo en un único componente llamado *tendencia-ciclo*, porque separarlos requiere información y supuestos adicionales. Una serie también puede presentar varias estacionalidades; en nuestros datos interesan especialmente la diaria y la anual.

| Componente | Qué representa | Cómo podría aparecer en los datos de Curicó |
| --- | --- | --- |
| **Tendencia (`T_t`)** | Evolución gradual del nivel de la serie a largo plazo. | Un aumento o una disminución sostenida de la temperatura media, que habría que comprobar en los registros. |
| **Estacionalidad (`S_t`)** | Patrones que se repiten con una periodicidad conocida. | Temperaturas generalmente más bajas cerca del amanecer y más altas por la tarde; inviernos más fríos y veranos más cálidos. |
| **Ciclo (`C_t`)** | Fluctuaciones cuya duración no es fija. | Períodos de varios años más cálidos o fríos de lo habitual, una vez considerado el patrón anual. |
| **Componente irregular (`e_t`)** | Variación que queda después de retirar los componentes anteriores. | Desviaciones asociadas al paso de un frente, cambios de nubosidad o viento, y errores de medición. |

**Estacionalidad y ciclo no son sinónimos.** El patrón anual se vincula con el calendario; un ciclo no tiene por qué repetirse cada cierto número exacto de años. Con los seis años disponibles, no intentaremos identificar un ciclo interanual separado de la tendencia. Tampoco interpretaremos una diferencia entre dos años como evidencia suficiente de una tendencia de largo plazo.

Además, **el componente irregular no equivale necesariamente a ruido impredecible**. Un descenso de temperatura que no se explica por la hora ni por la época del año podría anticiparse parcialmente con otras variables meteorológicas. Se llama irregular respecto de la descomposición utilizada.
</div>

<!-- DEFINICIÓN -->
<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: representación aditiva</strong>
En una **representación aditiva**, combinamos los componentes como una suma:

$$
y_t = T_t + S_t + C_t + e_t
$$

Cada componente aporta un cambio absoluto, expresado en las mismas unidades que la serie: por ejemplo, una desviación de −3 °C respecto del nivel esperado.
</div>


Para la **temperatura en grados Celsius**, usaremos esta representación para describir las diferencias respecto del nivel esperado.

  <strong style="display:block; margin-bottom:8px;">📖 Definición: baseline y climatología</strong>
Usaremos estos conceptos para diseñar un **baseline**, una regla sencilla que sirva como referencia al evaluar los modelos. En la Etapa 3 construiremos la regla llamada **climatología**: para cada pronóstico, usaremos el promedio histórico correspondiente a la **hora y al mes del instante que queremos predecir**, calculado únicamente con los datos de entrenamiento.

Por ejemplo, para pronosticar las 06:00 de una mañana de julio, la climatología usará el promedio de las temperaturas registradas a las 06:00 durante los meses de julio del entrenamiento. Así incorpora tanto el patrón diario como el anual. Un modelo más complejo tendrá que demostrar que mejora esa referencia: **no basta con obtener un error bajo si una regla basada en el calendario obtiene uno menor**.

Grafiquemos con [Plotly](https://plotly.com/python/) un origen concreto en el registro: las 18:00 del 15 de julio de 2020. El horizonte de 12 horas llega hasta las 06:00 del día siguiente. La parte sombreada muestra lo que ocurrió después: hoy podemos verla en el archivo histórico, pero al emitir el pronóstico todavía no estaba disponible.

````python
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

import plotly.express as px
import polars as pl

origen_ejemplo = datetime(2020, 7, 15, 18, tzinfo=ZoneInfo("America/Santiago"))
target_ejemplo = origen_ejemplo + timedelta(hours=12)
tramo_ejemplo = limpia.filter(
    pl.col("hora").is_between(
        origen_ejemplo - timedelta(hours=12), target_ejemplo
    )
)
pasado_ejemplo = tramo_ejemplo.filter(pl.col("hora") <= origen_ejemplo)
futuro_ejemplo = tramo_ejemplo.filter(pl.col("hora") >= origen_ejemplo)

figura_origen = px.line(
    pasado_ejemplo,
    x="hora",
    y="temp",
    markers=True,
    title="Origen a las 18:00 y target a las 06:00: horizonte de 12 horas",
    labels={"hora": "Hora local de Curicó", "temp": "Temperatura (°C)"},
)
figura_origen.update_traces(name="Disponible en el origen", showlegend=True)
figura_origen.add_scatter(
    x=futuro_ejemplo["hora"],
    y=futuro_ejemplo["temp"],
    mode="lines+markers",
    line={"dash": "dash", "color": "#b45309"},
    name="Futuro: visible solo al evaluar",
)
figura_origen.add_vrect(
    x0=origen_ejemplo,
    x1=target_ejemplo,
    fillcolor="#f59e0b",
    opacity=0.12,
    line_width=0,
    layer="below",
)
for hora_marcada, etiqueta in [
    (origen_ejemplo, "Origen: 18:00"),
    (target_ejemplo, "Target: 06:00"),
]:
    figura_origen.add_vline(x=hora_marcada, line_dash="dot")
    figura_origen.add_annotation(
        x=hora_marcada,
        y=1.05,
        yref="paper",
        text=etiqueta,
        showarrow=False,
        xanchor="right",
    )
figura_origen.update_layout(legend={"orientation": "h", "y": -0.25})
figura_origen.show()
````

Para pronosticar a las 18:00 no se puede usar la mínima de la noche: todavía no ocurrió. Se usan valores pasados de la temperatura (lags), resúmenes de las últimas horas (rolling windows), el calendario y features físicos que resumen el estado del aire.

<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: lag y rolling window</strong>
Un **lag** o **rezago** de orden `k` es el valor observado `k` pasos antes del origen: `temperatura(t − k)`. Como trabajamos sobre una grilla horaria completa, cada paso corresponde a una hora. El lag de orden 0 es la medición del propio origen y puede usarse si ya está disponible al pronosticar. Un desplazamiento hacia `t + 1` o `t + 2`, en cambio, trae información futura. Una **ventana móvil** (*rolling window*) resume varias observaciones consecutivas mediante una media, un mínimo, un máximo o una desviación estándar. Aquí las ventanas terminan en `t` e incluyen su medición: una ventana de 12 posiciones abarca desde `t − 11` hasta `t`. Una ventana centrada incluiría observaciones futuras y no serviría como feature para este pronóstico.

Estas transformaciones permiten representar la historia de la serie en una **tabla de aprendizaje supervisado** y trabajar con modelos de machine learning (ML) que reciben filas de variables predictoras, como LightGBM. Cada fila representa un origen `t`: sus columnas pueden contener la temperatura actual, la de 12 horas atrás y la media de las últimas 12 horas. Durante el entrenamiento, esa fila se asocia con un target: aquí, la temperatura observada en `t + 12`.

Los **lags** permiten comparar estados concretos del pasado; las **ventanas móviles** resumen el comportamiento reciente, como su nivel medio o su variabilidad. Un árbol puede aprender relaciones entre esas columnas y la temperatura futura: por ejemplo, que el pronóstico cambia según la temperatura actual y cuánto difiere de la media reciente. Para ello necesita que ese historial esté representado en las columnas que recibe.

Al emitir un pronóstico nuevo, construimos las mismas columnas con las mediciones disponibles hasta ese origen. El target se conocerá después. Aunque los ejemplos tengan forma de tabla, **la construcción de features y la separación entre entrenamiento y evaluación deben seguir respetando el tiempo**.
</div>


Dentro de una fila, el leakage ocurre cuando un feature de `t` usa información posterior a `t`. La forma más común es copiar el target como feature con `temp.shift(-h)`, que trae a la fila `t` el valor de `t + h`: en la tabla histórica la columna se ve completa, pero al pronosticar ese valor todavía no existe. Un rolling mean con `shift(-1)` o una rolling window centrada producen el mismo problema. Entre filas, el leakage aparece con un `train_test_split` con `shuffle=True` o con una normalización ajustada con todo el período. La regla es que **todo lo que se construya para la fila `t` debe poder calcularse con la información disponible en `t`**.


<!-- DEFINICIÓN -->
<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: punto de rocío, bulbo húmedo y componentes del viento</strong>
Estas definiciones son material de referencia. Todos sus cálculos vienen resueltos en `features.py`; no tienen que implementarlos.

El **punto de rocío** es la temperatura a la que una parcela de aire alcanza la saturación al enfriarse, manteniendo constantes la presión y su contenido de vapor de agua. La **depresión del punto de rocío**, `T − T_d`, indica cuánto tendría que enfriarse para alcanzar esa saturación: una diferencia pequeña indica aire próximo a saturarse. Al producirse condensación se libera calor, lo que puede moderar el enfriamiento; sin embargo, **el punto de rocío medido a las 18:00 no es un límite fijo para la mínima de la noche**, porque las condiciones del aire pueden cambiar. La **temperatura de bulbo húmedo** es la temperatura que registra un termómetro con el bulbo humedecido y ventilado, enfriado por evaporación; combina información de temperatura y humedad. Las **componentes del viento** representan su velocidad y dirección mediante dos valores: `u`, positivo hacia el este, y `v`, positivo hacia el norte. Así evitamos el salto numérico entre direcciones cercanas, como 359° y 1°.

  Con `T` en °C y `RH` en porcentaje, el punto de rocío se calcula con la fórmula de Magnus:

  $$\gamma = \ln(RH/100) + \frac{aT}{b+T}, \qquad T_d = \frac{b\gamma}{a-\gamma},$$

  con `a = 17,62` y `b = 243,12` °C. La depresión es `T - T_d`. Para `T = 20` °C y `RH = 50` %, `T_d = 9,26` °C.

  Con `w` en km/h y `theta` en grados desde el norte, las componentes del viento son

  $$u = -w \sin\theta, \qquad v = -w \cos\theta.$$

  `u` es la componente este-oeste y `v`, la norte-sur. El signo negativo aparece porque `theta` indica de dónde viene el viento: el viento del norte avanza hacia el sur, y por eso su `v` es negativo. Con viento del norte (`theta = 0`), `u = 0` y `v = -w`; con viento del este (`theta = 90`), `u = -w` y `v = 0`.
</div>

<!-- DEFINICIÓN -->
<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: ciclos de calendario</strong>
La **hora del día** y el **día del año** son variables de calendario con una estructura periódica: después de las 23:00 vienen las 00:00 y, después del último día del año, vuelve el primero. Una **codificación cíclica** representa cada valor mediante un par seno-coseno, de modo que los extremos del período queden próximos en esa representación. Puede ayudar al modelo a aprovechar esta continuidad, aunque no garantiza una mejora y los árboles también pueden trabajar con las variables originales. Con `h = hora + minuto/60` y `d = día del año`, usamos:

  $$\text{hora\_seno} = \sin\left(\frac{2\pi h}{24}\right), \qquad \text{hora\_coseno} = \cos\left(\frac{2\pi h}{24}\right)$$

  $$\text{dia\_anio\_seno} = \sin\left(\frac{2\pi (d-1)}{365.25}\right), \qquad \text{dia\_anio\_coseno} = \cos\left(\frac{2\pi (d-1)}{365.25}\right)$$

  El par `(seno, coseno)` recorre un círculo: las 00:00 quedan en `(0, 1)`, las 06:00 en `(1, 0)`, las 12:00 en `(0, -1)` y las 18:00 en `(-1, 0)`. Así las 23:00 y las 00:00 quedan cerca y el modelo puede representar cualquier fase del día. El `−1` del día del año hace que el 1 de enero empiece en fase 0. El denominador `365.25` aproxima la duración del año considerando los bisiestos; no representa exactamente la duración de cada año calendario.

  <p style="font-size:0.85em; margin-top:8px;">Fuente: Jean Fernandes, <a href="https://thejeanfernandes.medium.com/cyclical-time-encoding-the-hidden-trick-every-data-scientist-should-know-2f244a3ed45e">Cyclical Time Encoding: The Hidden Trick Every Data Scientist Should Know</a>, en <em>5-Minute Reads in Data Science and Machine Learning</em> (Medium).</p>

  En el círculo de la figura, el eje horizontal es el coseno y el vertical, el seno: sus coordenadas se leen como `(coseno, seno)`, en orden inverso al par de features descrito arriba. A la derecha se muestran las dos componentes durante el mismo ciclo, con `θ = 2πh/24`.

  <strong style="display:block; margin-bottom:8px;">📖 Definición: partición cronológica, holdout temporal y purga</strong>
Una **partición cronológica** separa los datos respetando su orden temporal. **Train** se usa para ajustar el modelo; **validation**, para comparar configuraciones y elegir decisiones como el umbral de alerta; **test**, para evaluar el procedimiento una vez fijadas esas decisiones. Un **holdout temporal** es un bloque de tiempo apartado del entrenamiento: puede utilizarse como validación o como prueba final, según el diseño de la evaluación.

La **purga** retira ejemplos de entrenamiento cuyos targets todavía no estarían disponibles al emitir los primeros pronósticos del bloque evaluado. Si el target es la temperatura `h` horas después del origen, debemos revisar los orígenes cercanos al corte; si resume una ventana futura, debemos considerar hasta cuándo se extiende esa ventana.

  <strong style="display:block; margin-bottom:8px;">📖 Definición: persistencia, persistencia de 24 horas y climatología</strong>

- **Persistencia** (columna `persistencia`): la temperatura de `t + h` será igual a la de `t`.
- **Persistencia de 24 horas** (columna `diaria`): la temperatura de `t + h` será igual a la de `t + h - 24`, la misma hora del target el día anterior. Es el «igual que ayer» de don Nicasio, y con `h` de 1 a 24 esa hora ya ocurrió en el origen.

<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: MAE y RMSE</strong>
El **error absoluto medio** (**MAE**) promedia la magnitud de los errores:

$$\mathrm{MAE} = \frac{1}{n}\sum_{i=1}^{n} |y_i - \hat{y}_i|$$

Un MAE de 1,7 °C significa que la diferencia absoluta entre la temperatura observada y la pronosticada fue, en promedio, de 1,7 °C. La **raíz del error cuadrático medio** (**RMSE**) se calcula como:

$$\mathrm{RMSE} = \sqrt{\frac{1}{n}\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$$

Al elevar los errores al cuadrado, da mayor peso a los errores grandes. Ambas métricas se expresan en °C y un valor menor indica mejores pronósticos sobre las mismas observaciones. El RMSE es mayor o igual que el MAE; una diferencia amplia indica mayor variación en la magnitud de los errores, pero hay que examinar su distribución para saber qué la produce. Ninguna de las dos distingue por sí sola entre sobreestimar y subestimar la temperatura. En las fórmulas, $n$ es el número de observaciones, $y_i$ la temperatura observada y $\hat{y}_i$ la pronosticada.
</div>

### Registro en MLflow

Desde esta etapa, cada resultado queda registrado como un run de MLflow: doña Eulalia va a querer saber de dónde salió cada número del análisis final, y una tabla del notebook se pierde cuando se vuelve a ejecutar.


<!-- DEFINICIÓN -->
<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: pasos y runs anidados</strong>
  En el Lab10 registraron cada evaluación como un <strong>run</strong>: los <strong>parámetros</strong> describen la configuración que se probó, las <strong>métricas</strong> son los resultados numéricos que se comparan y los <strong>tags</strong> clasifican el run para filtrarlo. Aquí suman dos ideas. Una métrica puede registrarse varias veces con un <strong>paso</strong> (<code>step</code>): <code>mlflow.log_metric("mae_horizonte", 2.2, step=12)</code> guarda el MAE a 12 horas, y la interfaz dibuja la curva de todos los pasos. Las tablas de runs muestran solo el último paso, así que el valor que se quiere comparar entre runs se registra además con su propio nombre. Un <strong>run anidado</strong> (<code>nested=True</code>) queda dentro de otro run, su padre, y sirve para agrupar, por ejemplo, los trials de una búsqueda bajo el run que la resume.   <strong style="display:block; margin-bottom:8px;">📖 Definición: boosting y LightGBM</strong>
El **boosting** construye un conjunto de modelos de manera secuencial: cada modelo nuevo busca mejorar el resultado del conjunto anterior. **LightGBM** implementa *gradient boosting* con árboles de decisión: en cada iteración agrega un árbol que busca reducir la función de pérdida del entrenamiento. Los **parámetros aprendidos** se obtienen al entrenar el modelo. En estos árboles incluyen las variables y los umbrales usados en las divisiones, y los valores de predicción de las hojas.   <strong style="display:block; margin-bottom:8px;">📖 Definición: predicciones fuera de muestra y MAE global</strong>
Las **predicciones fuera de muestra** de la validación (*out-of-fold*, OOF) se obtienen para observaciones que no participaron en el ajuste del modelo que las predice. En nuestro backtest, cada bloque de validación se predice con un modelo entrenado únicamente con información anterior a sus orígenes. El **MAE global** reúne los errores absolutos de todas esas predicciones y los promedia. Equivale a promediar los MAE de los folds ponderándolos por su número de observaciones:

$$\mathrm{MAE}_{\mathrm{global}} = \frac{\sum_j n_j\,\mathrm{MAE}_j}{\sum_j n_j}$$

Aquí, $n_j$ es el número de observaciones del fold $j$. Por eso, un fold con más observaciones pesa más; el resultado no necesariamente coincide con el promedio simple de los MAE por fold. Si usamos este valor para elegir hiperparámetros, sigue siendo una métrica de validación, no una evaluación final independiente.
</div>

  <strong style="display:block; margin-bottom:8px;">📖 Definición: Optuna, trial, función objetivo, estudio y sampler</strong>
**Optuna** es una biblioteca para buscar hiperparámetros. Un **trial** es una ejecución de la función objetivo que propone y evalúa una configuración. La **función objetivo** define cómo se evalúa esa configuración y devuelve el valor que queremos minimizar o maximizar; aquí devuelve el MAE global del backtest. Un **estudio** reúne los trials y conserva sus configuraciones y resultados. El **sampler** es el algoritmo que propone los valores de los hiperparámetros. Aquí usamos **TPE** (*Tree-structured Parzen Estimator*): tras una fase inicial de muestreo aleatorio, utiliza los resultados anteriores para favorecer configuraciones con mayor potencial de mejora. No garantiza encontrar el óptimo ni superar la configuración inicial.
</div>


La función objetivo corre el backtest sobre el período de desarrollo, con la purga anterior a test, y devuelve el MAE global de las predicciones que cada fold hace sobre su año de validation.

<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: umbral de alerta, recall y tasa de falsas alarmas</strong>
El **umbral de alerta** (`τ`) es el valor con el que comparamos la mínima pronosticada: emitimos una alerta si `mínima_pronosticada < τ`. Se elige en validation según los costos y se aplica a test sin modificarlo. Es distinto del umbral de 2 °C usado para definir el evento real. Un **verdadero positivo** (`TP`) es una helada avisada; un **falso negativo** (`FN`), una helada sin aviso; un **falso positivo** (`FP`), una alerta en una noche sin helada; y un **verdadero negativo** (`TN`), una noche sin helada ni alerta.

El **recall** o **sensibilidad** es `TP / (TP + FN)`: de todas las heladas reales, qué fracción se avisó. La **tasa de falsas alarmas**, entendida aquí como tasa de falsos positivos, es `FP / (FP + TN)`: de todas las noches sin helada, qué fracción tuvo alerta. No debe confundirse con `FP / (TP + FP)`, que mide la fracción de alertas que resultaron falsas.
</div>

Para elegir el umbral se recorren temperaturas candidatas en validation y se elige la de menor costo total: 400.000 pesos por cada noche con calefactores encendidos y 20.000.000 de pesos por cada helada sin aviso.


<!-- DEFINICIÓN -->
<div style="box-sizing:border-box; max-width:100%; overflow-wrap:break-word; padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición: costo esperado y umbral de probabilidad</strong>
El **costo esperado** promedia los costos de las posibles consecuencias de una decisión, ponderados por sus probabilidades. Para este ejercicio suponemos que encender los calefactores evita toda la pérdida atribuida a la helada y cuesta $C_a$, haya helada o no. No encender cuesta $C_h$ si ocurre una helada y cero si no ocurre.

Si $p$ es la probabilidad de helada dada la información disponible, el costo esperado de no encender es $p \times C_h$; el de encender es $C_a$. Por tanto, conviene encender cuando $p \times C_h > C_a$, es decir, cuando:

$$p > \frac{C_a}{C_h}$$

Con los costos del encargo, ese umbral es 0,02 o 2 %; en la igualdad, ambas decisiones tienen el mismo costo esperado. **Este es un umbral de probabilidad, no de temperatura.** Nuestro regresor predice una mínima, no una probabilidad de helada: el umbral de temperatura se elige comparando costos observados en validation, no convirtiendo directamente ese 2 % a grados Celsius.
</div>

La **importancia nativa** resume cómo se utilizaron las features al construir los árboles. En LightGBM puede medirse como **split**, el número de divisiones que usan una feature, o como **gain**, la suma de las ganancias de esas divisiones según el criterio de entrenamiento. La **importancia por permutación** mantiene fijo el modelo, desordena una columna de validation y calcula cuánto cambia el MAE respecto de usar los datos originales. Repetimos la permutación para resumir su variabilidad: un aumento grande indica que el modelo depende de esa información en el conjunto evaluado; un cambio cercano a cero o negativo requiere cautela al interpretarlo.

Los **valores SHAP** atribuyen a las features la diferencia entre una predicción y un **valor base**, definido por la referencia que utiliza el explicador: `predicción = valor_base + suma de aportes SHAP`. En este modelo de regresión, esos aportes se expresan en °C. Explican el comportamiento del modelo, no relaciones causales del sistema físico.

Si dos features contienen información parecida, la importancia puede repartirse entre ellas.
