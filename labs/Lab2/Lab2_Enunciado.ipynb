### ¿Qué es `main.py`?

`main.py` es el **punto de entrada** de su proyecto: el archivo que ejecutan cuando quieren correr el proyecto como programa, en vez de abrir un notebook. Ya tienen uno en la raíz de su repositorio — lo generó `uv init` en el Laboratorio 1 — y hasta ahora no lo habían tocado. Este es su contenido actual:

```python
def main():
    print("Hello from mds7202-v2!")


if __name__ == "__main__":
    main()
```

La parte que probablemente no reconocen es `if __name__ == "__main__":`.

`__name__` es una variable especial que Python define automáticamente en cada archivo. Vale `"__main__"` cuando ese archivo es el que se ejecutó directamente, y vale el nombre del módulo cuando el archivo fue importado desde otro lugar.

Ese `if` aprovecha la diferencia para separar dos cosas: el código que **define** funciones y clases, que debe estar disponible siempre, del código que **se ejecuta** al correr el programa, que solo debería dispararse cuando alguien ejecuta ese archivo a propósito.

### Atributos, constructor y métodos

El constructor permite crear autos con un estado inicial. En este ejemplo
guardamos la marca, el modelo, la velocidad máxima y la velocidad actual.


<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición — constructor, atributos y métodos</strong>
  El <strong>constructor</strong> es el método <code>__init__</code>, que se ejecuta al crear un objeto y deja listo su estado inicial. Un <strong>atributo</strong> es un dato asociado al objeto, como <code>marca</code>. Un <strong>método</strong> es una función definida dentro de la clase, como <code>acelerar</code>, que puede leer o modificar ese estado.
</div>

````python
class Auto:
    def __init__(
        self,
        marca: str,
        modelo: str,
        velocidad_maxima: float,
    ) -> None:
        self.marca = marca
        self.modelo = modelo
        self.velocidad_maxima = velocidad_maxima
        self._velocidad = 0.0

    def acelerar(self, incremento: float) -> None:
        self._velocidad += incremento

    def frenar(self, decremento: float) -> None:
        self._velocidad -= decremento


auto = Auto("Toyota", "Yaris", 180)
auto.acelerar(30)
auto.frenar(10)

print(auto.marca, auto.modelo)
print(auto.velocidad_maxima)
print(auto._velocidad)
````

No impiden por sí solos que alguien entregue otro tipo de dato, pero hacen más
clara la interfaz del objeto.


<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición — type hints</strong>
  Los <strong>type hints</strong> son anotaciones que indican los tipos esperados para parámetros y retornos. Sirven como documentación y ayudan a las herramientas de desarrollo, pero Python no los convierte automáticamente en validaciones durante la ejecución.
</div>

## Herencia

La herencia permite crear clases nuevas a partir de una clase base. Las clases
derivadas reutilizan los atributos y métodos comunes, y pueden agregar o
sobrescribir comportamientos.


<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición — herencia y <code>super()</code></strong>
  La <strong>herencia</strong> permite que una subclase reutilice la estructura y los comportamientos de una clase base. <code>super()</code> permite acceder a la implementación de la clase base, especialmente para ejecutar su constructor, sin repetir ese código.
</div>

````python
class AutoElectrico(Auto):
    def fuente_energia(self) -> str:
        return "batería"


class AutoCombustion(Auto):
    def fuente_energia(self) -> str:
        return "combustible"


electrico = AutoElectrico("Tesla", "Model 3", 200)
combustion = AutoCombustion("Suzuki", "Swift", 180)

print(electrico.fuente_energia())
print(combustion.fuente_energia())
````

  Las clases derivadas reutilizan lo común de Auto, pero cada una puede definir su propia fuente de energía.

La clase base concentra lo que todos los autos comparten. Una subclase

## Abstracción y encapsulación

Una clase abstracta define la interfaz que sus subclases deben cumplir. Para
los autos, toda variante debe informar cuál es su fuente de energía, pero la
clase base no puede decidir esa respuesta por sí sola.


<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición — abstracción, encapsulación y propiedad</strong>
  La <strong>abstracción</strong> define qué debe ofrecer un objeto sin obligar a mostrar cómo lo implementa. La <strong>encapsulación</strong> reúne el estado y sus operaciones, y recomienda no depender directamente de los detalles internos. Una <strong>propiedad</strong> permite consultar o modificar un atributo mediante una interfaz como si fuera un atributo, manteniendo la lógica dentro de la clase.
</div>

````python
from abc import ABC, abstractmethod


class Auto(ABC):
    def __init__(
        self,
        marca: str,
        modelo: str,
        velocidad_maxima: float,
    ) -> None:
        self.marca = marca
        self.modelo = modelo
        self.velocidad_maxima = velocidad_maxima
        self._velocidad = 0.0

    @abstractmethod
    def fuente_energia(self) -> str: ...

    @property
    def velocidad(self) -> float:
        return self._velocidad

    def acelerar(self, incremento: float) -> None:
        self._velocidad += incremento

    def frenar(self, decremento: float) -> None:
        self._velocidad -= decremento


class AutoElectrico(Auto):
    def fuente_energia(self) -> str:
        return "batería"


class AutoCombustion(Auto):
    def fuente_energia(self) -> str:
        return "combustible"


electrico = AutoElectrico("Tesla", "Model 3", 200)
combustion = AutoCombustion("Suzuki", "Swift", 180)

electrico.acelerar(30)
combustion.acelerar(30)

print(electrico.velocidad)
print(combustion.velocidad)
````

<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #d29922; background:rgba(210,153,34,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">⚠️ Aviso</strong>
  Auto no se puede instanciar porque tiene un método abstracto. Las subclases sí se pueden crear cuando implementan fuente_energia.
</div>


````python
try:
    Auto("Toyota", "Yaris", 180)
except TypeError as error:
    print(f"No se puede crear Auto: {error}")
````

La propiedad velocidad entrega una forma simple de consultar el estado
del auto sin exponer directamente cómo se guarda internamente. La convención
_protegida no es una barrera de seguridad: es una señal para quien usa la clase.


<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición — atributos protegidos y privados</strong>
  En Python, un atributo con un prefijo <code>_</code> se considera <strong>protegido por convención</strong>: se puede acceder, pero el código externo debería tratarlo como un detalle interno. Un prefijo doble, como <code>__velocidad</code>, activa un mecanismo de renombrado interno (*name mangling*); tampoco constituye una barrera de seguridad.
</div>

## Polimorfismo

Una función polimórfica trabaja con objetos distintos siempre que todos
respondan al método que la función necesita.


<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición — polimorfismo</strong>
  El <strong>polimorfismo</strong> permite que una misma función trabaje con objetos de clases distintas mediante una interfaz común. La función se concentra en qué operación necesita, no en preguntar de qué tipo concreto es cada objeto.
</div>

````python
def describir_fuentes(autos: list[Auto]) -> None:
    for auto in autos:
        print(f"{auto.modelo}: {auto.fuente_energia()}")


autos = [electrico, combustion]
describir_fuentes(autos)
````

## Excepciones

Una excepción permite comunicar que una operación no puede completarse
correctamente. Seguiremos usando Auto para distinguir una velocidad válida de
una velocidad imposible.


<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición — excepción, <code>raise</code> y <code>try/except</code></strong>
  Una <strong>excepción</strong> es un objeto que comunica que ocurrió una situación que impide continuar normalmente. <code>raise</code> la lanza; <code>try</code> delimita el código que podría fallar y <code>except</code> define cómo responder a un tipo de error específico.
</div>

````python
class VelocidadInvalidaError(Exception):
    pass


def validar_velocidad(auto: Auto, velocidad: float) -> None:
    if velocidad < 0 or velocidad > auto.velocidad_maxima:
        raise VelocidadInvalidaError(
            f"La velocidad debe estar entre 0 y {auto.velocidad_maxima} km/h"
        )


electrico = AutoElectrico("Tesla", "Model 3", 200)

try:
    validar_velocidad(electrico, 250)
except VelocidadInvalidaError as error:
    print(f"Velocidad inválida: {error}")
````

  Una excepción distingue entre una operación válida y una operación que no puede producir un resultado confiable.

## Testing y testing unitario

El <strong>testing</strong> consiste en ejecutar comprobaciones para buscar errores y confirmar que el programa cumple el comportamiento esperado. En este laboratorio usaremos <code>pytest</code> para descubrir y ejecutar las pruebas ubicadas en <code>tests/</code>.

El <strong>testing unitario</strong> verifica una pieza pequeña y aislada del programa, por ejemplo, un método de una clase o una función. Una prueba debe tener un caso claro, una acción y una comprobación del resultado.

  <strong style="display:block; margin-bottom:8px;">📖 Definición — testing unitario</strong>
  Una <strong>prueba unitaria</strong> comprueba un comportamiento pequeño sin depender de todo el sistema. Por ejemplo, podemos probar por separado que un auto eléctrico informe "batería" o que un sensor de temperatura reconozca una situación riesgosa.

  <strong style="display:block; margin-bottom:8px;">📖 Definición — <code>assert</code></strong>
  <code>assert</code> verifica que una expresión sea verdadera. Si lo es, la prueba continúa; si es falsa, Python lanza <code>AssertionError</code> y pytest marca la prueba como fallida. La condición debe expresar el resultado que esperamos observar.
</div>

````python
auto_prueba = AutoElectrico("Tesla", "Model 3", 200)

assert auto_prueba.fuente_energia() == "batería"
auto_prueba.acelerar(30)
assert auto_prueba.velocidad == 30

print("Las comprobaciones pasaron")
````

### Una prueba unitaria con pytest

Las pruebas se escriben como funciones cuyo nombre comienza con <code>test_</code>. Dentro de ellas, <code>assert</code> expresa el resultado esperado. Al ejecutar <code>uv run pytest</code>, pytest busca automáticamente esos archivos y funciones.

En AgroAlerta, por ejemplo, una prueba unitaria tendrá esta forma:

```python
def test_temperatura_bajo_cero_es_riesgo():
    sensor = SensorTemperatura(0, 40)

    assert sensor.es_riesgo(-2) is True
```

Para comprobar que una operación levanta una excepción se puede usar <code>pytest.raises</code>. Con el ejemplo de los autos:

```python
def test_velocidad_sobre_el_maximo_levanta_error():
    auto = AutoElectrico("Tesla", "Model 3", 200)

    with pytest.raises(VelocidadInvalidaError):
        validar_velocidad(auto, 250)
```

La prueba no debe limitarse a ejecutar el código: debe afirmar qué resultado o excepción se esperaba.
