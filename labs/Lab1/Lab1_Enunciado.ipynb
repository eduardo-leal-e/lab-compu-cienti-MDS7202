## Jupyter

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición</strong>
  Jupyter es un entorno interactivo para combinar texto, código y resultados en un mismo documento. Un notebook está formado por celdas y utiliza un kernel para ejecutar el código. Las celdas Markdown contienen texto informativo y las celdas de código permiten ejecutar pequeños experimentos paso a paso.
</div>

En ciencia de datos, un notebook puede servir para explorar datos, documentar un experimento, mostrar resultados y dejar una explicación reproducible. El notebook no reemplaza necesariamente a un módulo de código, pero ayuda a comunicar el razonamiento detrás del trabajo.

## Usar un notebook en VS Code

Para trabajar con este archivo en VS Code:

1. Instala la extensión **Jupyter** de Microsoft.
2. Abre el archivo `.ipynb`.

- El **modo edición** permite escribir dentro de una celda. 
- El **modo comando** permite seleccionar celdas y ejecutar acciones sobre ellas. 


Presiona `Esc` para pasar al modo comando y `Enter` para volver a editar la celda seleccionada.

## Atajos útiles para notebooks

| Atajo | Acción |
|---|---|
| `Ctrl + Enter` | Ejecutar la celda manteniendo la selección |
| `Alt + Enter` | Ejecutar la celda e insertar una nueva debajo |
| `Esc` | Pasar al modo comando |
| `Enter` | Editar la celda seleccionada |
| `A` | Insertar una celda arriba |
| `B` | Insertar una celda abajo |
| `M` | Convertir la celda a Markdown |
| `Y` | Convertir la celda a código |
| `DD` | Eliminar la celda seleccionada |
| `Z` | Deshacer la eliminación de una celda |

En macOS, usa `Cmd` cuando el atajo indique `Ctrl`. 

Comprueba los atajos disponibles desde la paleta de comandos de VS Code si tu configuración es distinta.

# Sección 1: GitHub y Git 👾

## Git

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición</strong>
  Git es un sistema distribuido de control de versiones. Registra cambios en un historial de commits y permite trabajar en ramas separadas antes de integrar los cambios. Cada integrante puede tener una copia local del historial y sincronizarla con uno o más repositorios remotos.
</div>

## Repositorio

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición</strong>
  Un repositorio es una carpeta administrada por Git que contiene los archivos del proyecto y el historial de cambios. Puede ser local, cuando está en tu equipo, o remoto, cuando está alojado en un servicio como GitHub.
</div>

Un repositorio no es solamente una carpeta compartida: también conserva quién cambió cada archivo, cuándo lo hizo y qué diferencias introdujo.

## Markdown para documentación técnica

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición</strong>
  Markdown es un lenguaje de marcado liviano que permite escribir texto estructurado usando caracteres simples. Luego, ese texto puede convertirse en títulos, listas, enlaces, tablas, imágenes y bloques de código.
</div>

En ciencia de datos aparece en los `README.md`, notebooks, reportes de experimentos, documentación de datasets y Pull Requests. La idea es que otras personas puedan entender qué hace un proyecto, cómo ejecutarlo y qué decisiones se tomaron.

### Elementos frecuentes de Markdown

| Elemento | Escritura | Uso |
|---|---|---|
| Título | `# Título` | Organizar secciones |
| Énfasis | `**importante**` | Destacar información |
| Lista | `- elemento` | Enumerar ideas o pasos |
| Enlace | `[texto](https://ejemplo.com)` | Referenciar recursos |
| Imagen | `![alt](imagen.png)` | Incorporar una imagen |
| Código | `git status` | Mostrar un comando |
| Bloque de código | Tres backticks | Mostrar un script completo |
| Tabla | Filas separadas por `|` | Comparar información |

El texto que escribes es la fuente Markdown; la vista previa muestra su representación renderizada. Los alerts de este laboratorio combinan una celda Markdown con HTML para controlar mejor su presentación visual.

### Parte 1: Creando y actualizando una presentación personal (1.5 pts)

Para empezar a soltar las manos con Git y GitHub, crea un repositorio personal con una presentación. El repositorio debe llevar tu nombre de usuario.

El repositorio debe contener un breve resumen de quién eres, tus gustos, experiencias y cualquier otro detalle que consideres relevante.

### Crear el repositorio personal

Cuenta con una cuenta de GitHub y ejecuta:

```bash
gh repo create [usuario] --public --add-readme --clone
cd [usuario]
```

Reemplaza `[usuario]` por tu nombre de usuario real.

### Subir los cambios y abrir el Pull Request

Registra los cambios y sube la rama:

```bash
git add README.md
git commit -m "feat: agregar perfil personal"
git push -u origin perfil-inicial
```

Luego abre el Pull Request desde la interfaz gráfica de GitHub:

1. Entra al repositorio personal.
2. Selecciona **Compare & pull request**.
3. Elige `main` como branch base y `perfil-inicial` como branch de trabajo.
4. Completa el template de Pull Request.
5. Crea y mezcla el Pull Request.

La información del perfil debe llegar a `main` mediante el Pull Request, no mediante un commit directo.

### Crear la rama y editar el perfil

Crea una rama para actualizar el perfil y modifica `README.md` con la información que desees compartir:

```bash
git checkout -b perfil-inicial
```

Puedes guiarte por esta [presentación](https://github.com/pbadillatorrealba/pbadillatorrealba) o por los templates de [Awesome GitHub Profile README](https://durgeshsamariya.github.io/awesome-github-profile-readme-templates/#/).

Incluye títulos, listas, enlaces o imágenes de Markdown que ayuden a presentar tu perfil.

## `uv`

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición</strong>
  `uv` es una herramienta para crear entornos, resolver dependencias, mantener archivos de lock y ejecutar comandos dentro del entorno del proyecto.
</div>

| Comando | Propósito |
|---|---|
| `uv sync` | Sincronizar el entorno con la configuración del proyecto |
| `uv run` | Ejecutar un comando dentro del entorno del proyecto |
| `uv add` | Agregar una dependencia y actualizar la configuración |
| `uv lock` | Resolver y actualizar `uv.lock` |

En este laboratorio se usará `uv sync --locked --all-groups`: `--locked` exige que el lock sea consistente y `--all-groups` instala también las herramientas de desarrollo.

# Sección 2: Práctica de Git 👾

## Git: modelo mental

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63,185,80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición</strong>
  Git trabaja con snapshots y un historial de commits. Una rama es un puntero a una línea de trabajo y `HEAD` indica dónde estás trabajando actualmente.
</div>

Para entender qué ocurre al ejecutar un comando, distingue cuatro espacios:

```text
working tree --git add--> staging area --git commit--> repositorio local
      ^                         ^                         |
      |                         |                         | git push
      +------ git restore       +------ git reset         v
                                                   repositorio remoto
```

El working tree contiene tus archivos actuales. El staging area contiene los cambios que seleccionaste para el próximo commit. El repositorio local contiene el historial y el remoto permite compartirlo.

Git es útil para versionar código, configuraciones, notebooks y documentación de un proyecto de datos. No reemplaza un sistema de almacenamiento para datasets grandes ni debe usarse para guardar secretos, ambientes virtuales, logs o modelos pesados.

## Operaciones principales

| Operación | Para qué sirve |
|---|---|
| `git status` | Revisar branch, cambios y conflictos |
| `git diff` | Inspeccionar cambios todavía no preparados |
| `git add` | Pasar cambios al staging area |
| `git commit` | Guardar un snapshot en el historial local |
| `git log` | Inspeccionar commits, ramas y relaciones |
| `git branch` | Crear o listar branches |
| `git checkout` | Cambiar de branch o recuperar archivos |
| `git fetch` | Descargar referencias remotas sin modificar tu branch |
| `git pull` | Descargar cambios e integrarlos en la branch actual |
| `git push` | Publicar commits locales en un remoto |
| `git merge` | Integrar el historial de otra branch |
| `git merge --abort` | Cancelar un merge con conflictos |
| `git blame` | Ver qué commit modificó cada línea |
| `git restore` | Recuperar o descartar cambios de archivos |
| `git stash` | Guardar temporalmente cambios sin commit |

<!-- TIP -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #2f81f7; background:rgba(47,129,247,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">💡 Tip</strong>
  Ejecuta `git status` antes y después de cada operación importante. Los commits pequeños y descriptivos facilitan revisar experimentos, revertir errores y entender la historia del proyecto.
</div>

<!-- IMPORTANTE -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #8957e5; background:rgba(137,87,229,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📌 Importante</strong>
  `origin/fix-ifs` es una referencia a una branch del remoto `origin`. Después de clonar, puede no existir una branch local llamada `fix-ifs`; por eso se usa explícitamente `git merge origin/fix-ifs`.
</div>

<!-- CAUTION -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #f85149; background:rgba(248,81,73,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">🛑 Caution</strong>
  No ejecutes `git reset --hard`, `git clean` o `git push --force` sin entender qué cambios eliminarás o sobrescribirás. En branches compartidas puedes borrar trabajo de otras personas.
</div>
