## Etapa 3 — Operadores con saturación y métodos especiales (1,5 puntos)

<!-- DEFINICIÓN -->
<div style="padding:16px 20px; margin:16px 0; border-left:4px solid #3fb950; background:rgba(63, 185, 80,.12); color:inherit; border-radius:4px;">

  <strong style="display:block; margin-bottom:8px;">📖 Definición — métodos especiales o dunder methods</strong>
  En Python, los métodos cuyos nombres comienzan y terminan con dos guiones bajos son <strong>métodos especiales</strong> o <em>dunder methods</em>. Python los invoca automáticamente cuando usamos una operación o una sintaxis asociada. Por ejemplo, la expresión <code>imagen + 50</code> busca ejecutar <code>imagen.__add__(50)</code>.
  
  <pre><code>import numpy as np
from src.pixellab.imagen import Imagen

imagen = Imagen(np.array([[[100, 120, 140]]], dtype=int))
resultado = imagen + 50
print(resultado.imagen)
# [[ [150 170 190] ]]

# En sentido inverso, Python puede usar __radd__:
resultado = 50 + imagen</code></pre>
  
  La estructura de un método especial puede verse así. El cuerpo queda como
  un stub para que ustedes lo completen:
  <pre><code>class Imagen:
    def __add__(self, other):
        # Su código aquí
        raise NotImplementedError(
            "Completen __add__ antes de ejecutar esta celda."
        )</code></pre>
  
  En este laboratorio implementarán métodos especiales para que <code>Imagen</code> se comporte como un arreglo frente a escalares, arreglos de NumPy y otras imágenes, sin perder la validación ni la saturación de las intensidades.
</div>

Implementen en `Imagen` los siguientes operadores. Cada uno debe poder
trabajar con `int | float | np.ndarray | Imagen`:

- `__add__(self, other)`: define `imagen + other`; suma el operando a cada
  intensidad de la imagen.
- `__radd__(self, other)`: define `other + imagen`; como la suma es
  conmutativa, puede delegar en `__add__`.
- `__sub__(self, other)`: define `imagen - other`; resta el operando a cada
  intensidad de la imagen.
- `__rsub__(self, other)`: define `other - imagen`; debe respetar este orden
  porque la resta no es conmutativa.
- `__mul__(self, other)`: define `imagen * other`; multiplica cada
  intensidad por el operando.
- `__rmul__(self, other)`: define `other * imagen`; como la multiplicación
  es conmutativa, puede delegar en `__mul__`.

En todos los casos:

- si `other` es una `Imagen`, validen que sus dimensiones coincidan; si no,
  lancen `ValueError` con el mensaje «Las dimensiones de la imagen a operar
  (alto x ancho x canales) no calzan con las de la imagen original (alto x
  ancho x canales)»;
- conviertan el resultado a `int` con `.astype(int)`, incluso cuando
  `other` sea flotante;
- saturen mediante indexado condicional al rango `[0, 255]`;
- retornen un objeto `Imagen` nuevo con una copia (`np.copy`) del arreglo;
- no muten `self.imagen`.

## Etapa 4 — Procesamiento I: negativo, gris y canal (0,9 puntos)

En `src/pixellab/procesamiento.py`, creen la clase `LibImagen`. Cada método recibe una `Imagen` y retorna una `Imagen` nueva, con `.astype(int)` y sin mutar la entrada:

- `to_negative(img_in)`: `255 - img`;

## Etapa 5 — Procesamiento II: flip, saturación y contraste (1,1 puntos)

Agreguen a `LibImagen`:

- `flip(img_in, axis)`: `'h'` invierte las columnas y `'v'` las filas; con un eje inválido lanza `ValueError` con «Eje 'x' no válido. Valores posibles: 'h' (horizontal) o 'v' (vertical).»;
- `set_saturation(img_in, C)`: `R = gris + C * (img - gris)`, reutilizando `to_gray`; con `C > 1` aumenta la saturación, `C < 1` la reduce y `C == 0` produce escala de grises;
- `set_contrast(img_in, C)`: con `F = 259 * (C + 255) / (255 * (259 - C))` y `R = F * (img - 128) + 128`.

Los tres usan `.astype(int)` y saturan a `[0, 255]` con indexado condicional.
